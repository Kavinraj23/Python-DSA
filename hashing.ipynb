```python
# Hashsets
s = set()
print(s)
```

```python
# add item into set - O(1)
s.add(1)
s.add(2)

print(s)
```

```python
# lookup if item is in set - O(1)
print(1 in s)  # True
```

```python
print(1 not in s)
```

```python
# remove item from set - O(1)
s.remove(1) # O(1)
```

```python
s
```

```python
string = 'aaaaaaaaaaaabbbbbbdbbbbbbbbbbbbbccccccccccc'
s = set(string) # Set construction - O(S) - S is the length of the string
s
```

```python
# Loop over items in set - O(N) - N is the number of items in the set
for x in s:
    print(x)
```

```python
# Hashmaps - Dictionaries

d = {'greg':1, 'sally':2, 'bob':3}
d
```

```python
# add key-value in dictionary - O(1)
d['arsh'] = 4
print(d)
```

```python
# Check for presence of key - O(1)
if 'greg' in d:
    print(True)
```

```python
# check the value corresponding to a key - O(1)
print(d['greg']) # 1
```

```python
# Loop over key-value pairs - O(N) - N is the number of items in the dictionary
for k, v in d.items():
    print(k, v)
```

```python
# Defaultdict - assigns a default value to a key if it does not exist in the dictionary
from collections import defaultdict
d = defaultdict(int)
d[2]
```
