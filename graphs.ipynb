```python
# Array of Edges (Directed) [start, end]
n = 8
A = [[0, 1], [1, 2], [0, 3], [3, 4], [3, 6], [3, 7], [4, 2], [4, 5], [5, 2]]
```

```python
A
```

```python
# Convert Array of Edges -> Adjacency Matrix
M = []
for i in range(n):
    M.append([0] * n)

for u, v in A:
    M[u][v] = 1

    # uncomment the follow line if the graph is undirected
    # M[v][u] = 1

M
```

```python
# Convert Array of Edges -> Adjacency List
from collections import defaultdict

D = defaultdict(list)

for u, v in A:
    D[u].append(v)
    # uncomment the follow line if the graph is undirected
    # D[v].append(u)

D
```

```python
D[3]
```

```python
# DFS with Recursion - O(V + E)

def dfs_recursive(node):
    print(node)
    for nei_node in D[node]:
        if nei_node not in seen:
            seen.add(nei_node)
            dfs_recursive(nei_node)

source = 0
seen = set()
seen.add(source)
dfs_recursive(source)
```

```python
# DFS Iterative with Stack - O(V + E)
# dfs follows a path as far as it can go before backtracking

source = 0
seen = set()
seen.add(source)
stk = [source]
while stk:
    node = stk.pop()
    print(node)
    for nei_node in D[node]:
        if nei_node not in seen:
            seen.add(nei_node)
            stk.append(nei_node)
```

```python
# BFS (Queue) - O(V + E)
# bfs explores all neighbors at current depth before moving deeper

source = 0

from collections import deque

seen = set()
seen.add(source)
q = deque()
q.append(source)

while q:
    node = q.popleft()
    print(node)
    for nei_node in D[node]:
        if nei_node not in seen:
            seen.add(nei_node)
            q.append(nei_node)
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.neighbors = []
    
    def __str__(self):
        return "Node({})".format(self.value)
    
    def display(self):
        connections = [node.value for node in self.neighbors]
        return "{} is connected to: {}".format(self.value, connections)
    
A = Node('A')
B = Node('B')
C = Node('C')
D = Node('D')

A.neighbors.append(B)
B.neighbors.append(A)

C.neighbors.append(D)
D.neighbors.append(C)
```
